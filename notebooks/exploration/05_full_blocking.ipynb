```python
import sys, time, gc
from pathlib import Path
import duckdb, pandas as pd

ROOT = Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import src.blocking as B
import src.rerank as R

ARTIFACTS = ROOT / "artifacts"
KEYS2 = ARTIFACTS / "keys_v2"
def k2(name): return f"read_parquet('{(KEYS2 / f'{name}.parquet').as_posix()}')"

con = duckdb.connect((ARTIFACTS / "blocking.duckdb").as_posix())
B.setup(con, threads=4, memory_limit="4GB", temp_dir=(ARTIFACTS / "duckdb_tmp").as_posix())

S1_ALL    = k2("train_source1")
TGT_PARTS = [k2("train_source2"), k2("train_source3")]
TGT       = "(" + " UNION ALL ".join(f"SELECT * FROM {p}" for p in TGT_PARTS) + ")"

booster = R.load(ARTIFACTS / "rerank_lgb.txt")
print(sys.executable)
print(con.execute("SELECT (SELECT count(*) FROM eval_s1), (SELECT count(*) FROM eval_pairs)").fetchone())
```

```python
con.execute("CREATE OR REPLACE TABLE cut_full (s1_id BIGINT, tgt_id BIGINT, src BIGINT, p DOUBLE, rk2 BIGINT)")
t0 = time.time()
B.run_blocking(con, "cut_full", S1_ALL, TGT_PARTS, s1_ids_tbl=None, k_max=30, s1_chunks=32,
               on_chunk=lambda tbl: R.rerank_and_cut(con, tbl, S1_ALL, TGT, booster, "cut_full", tau=0.01, cap=5))
print(f"done in {(time.time() - t0) / 60:.1f} min")
```

```python
import src.metrics as M

con.execute("""
CREATE OR REPLACE TABLE cut_rev AS
SELECT *,
       row_number() OVER (PARTITION BY tgt_id ORDER BY p DESC, s1_id) AS rev_rk,     -- 1 = this S1 is the target's best claimant
       count(*)     OVER (PARTITION BY tgt_id)                         AS n_claim,    -- how many S1 compete for this target
       max(p)       OVER (PARTITION BY tgt_id)                         AS tgt_max_p   -- strongest claim on this target
FROM cut_full
""")

print(con.execute("""
SELECT avg((n_claim > 1)::INT) AS share_contested, avg(n_claim) AS avg_claimants
FROM cut_rev
""").fetchdf())

# Singleton hypothesis: for eval singletons, is their best candidate claimed more strongly by another S1?
display(con.execute("""
WITH top AS (
    SELECT r.*, row_number() OVER (PARTITION BY r.s1_id ORDER BY r.p DESC) AS own_rk
    FROM cut_rev r SEMI JOIN eval_s1 e USING (s1_id)
),
lab AS (
    SELECT t.*, (e.matched_entity_ids IS NULL OR trim(e.matched_entity_ids) = '') AS is_singleton
    FROM top t JOIN eval_s1 e USING (s1_id) WHERE own_rk = 1
)
SELECT is_singleton, count(*) AS n,
       avg(p)                       AS avg_top_p,
       avg((rev_rk > 1)::INT)       AS top_cand_owned_by_other_s1,
       avg(tgt_max_p - p)           AS avg_gap_to_best_claimant
FROM lab GROUP BY 1
""").fetchdf())

rows = []
for t in (0.5, 0.6, 0.7, 0.8):
    rows.append(M.eval_f05(con, f"(SELECT s1_id, tgt_id FROM cut_rev WHERE p >= {t})", f"p>={t}"))
    rows.append(M.eval_f05(con, f"(SELECT s1_id, tgt_id FROM cut_rev WHERE p >= {t} AND rev_rk = 1)",
                           f"p>={t} + one-to-many"))
display(pd.concat(rows, ignore_index=True))
```
